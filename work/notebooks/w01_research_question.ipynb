## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Looking at the starter dataset, there is a massive opportunity for prioritization. Out of 30,000 pages, many are actively trending downwards, but only a fraction actually have enough traffic to matter. Filtering the noise and finding the true opportunities is essential because a human team cannot review thousands of pages manually.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the starter data directly from the GitHub raw URL
url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

total_pages = len(df)
declining_pages = len(df[df['trend_direction'] == 'down'])
high_imp_declining = len(df[(df['trend_direction'] == 'down') & (df['impressions_90d'] >= 500)])

print(f"Total pages in starter slice: {total_pages}")
print(f"Pages trending down: {declining_pages}")
print(f"Trending down WITH substantial visibility (>= 500 impressions): {high_imp_declining}")
print("Conclusion: We need a model to rank these candidates because human teams cannot manually review thousands of declining pages.")
```
