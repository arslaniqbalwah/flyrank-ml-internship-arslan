## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

For the starter dataset, the target is a proxy label: a defined rule where trend_direction == "down". While this helps build the initial pipeline, a stronger future label for the capstone would be an observed outcome, such as predicting an actual drop in traffic over a future 30-day window based on the prior 90 days of features.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Define the proxy target from the starter dataset
df['is_declining'] = df['trend_direction'] == 'down'

print("Target variable distribution (Proxy Label):")
print(df['is_declining'].value_counts(normalize=True).round(2) * 100)
```

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

print("Success Metric: Precision@K (e.g., Precision@50).")
print("Why: Evaluates the quality of the top-ranked recommendations, matching real-world editorial capacity.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (like an if-statement checking impressions > 500 AND age > 180) is too rigid and treats all metrics independently. ML beats a fixed rule because it can learn complex, non-linear interactions—for example, how a slight drop in average position interacts dynamically with search volume, CTR, and freshness all at once to identify nuanced decay patterns that strict thresholds miss.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

print("Advantage of ML over fixed rules:")
print("- Fixed rules use hard cutoffs (e.g., age > 180).")
print("- ML learns continuous, non-linear combinations of age, volume, and engagement.")
```
